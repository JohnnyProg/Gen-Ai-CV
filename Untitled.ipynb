```python
import glob, os
def readfiles(path):
    pdfs = []
    for file in glob.glob(os.path.join(path, "*.pdf")):
        print(file)
        pdfs.append(file)
    return pdfs

pdfs = readfiles("./resumes_dataset/resumes_dataset/")
```

```python
import PyPDF2
def openFile(file_path):
    text = ""
    with open(file_path, "rb") as f:
        reader = PyPDF2.PdfReader(f)
        for page in reader.pages:
            text = text + "\n" + page.extract_text()
    return text
print(openFile(pdfs[0]))
```
